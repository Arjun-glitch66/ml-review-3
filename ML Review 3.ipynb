```python

weather_columns=[
    "DATE_TIME",
    "PLANT_ID",
    "AMBIENT_TEMPERATURE",
    "MODULE_TEMPERATURE",
    "IRRADIATION"
]

df_plant_1=pd.merge(
    df_generation_1,
    df_weather_1[weather_columns],
    on=["DATE_TIME","PLANT_ID"],
    how="left",
    validate="many_to_one",
    indicator=True
)

df_plant_2=pd.merge(
    df_generation_2,
    df_weather_2[weather_columns],
    on=["DATE_TIME","PLANT_ID"],
    how="left",
    validate="many_to_one",
    indicator=True
)

print("Plant 1 merge results:")
print(df_plant_1["_merge"].value_counts())

print("\nPlant 2 merge results:")
print(df_plant_2["_merge"].value_counts())

df_plant_1=df_plant_1.drop(columns="_merge")
df_plant_2=df_plant_2.drop(columns="_merge")

print("\nPlant 1 shape:",df_plant_1.shape)
print("Plant 2 shape:",df_plant_2.shape)
```

```python

df=pd.concat(
    [df_plant_1,df_plant_2],
    ignore_index=True
)

df=df.sort_values(
    ["PLANT_ID","DATE_TIME","SOURCE_KEY"]
).reset_index(drop=True)

print("Combined dataset shape:",df.shape)
print("\nFirst five records:")
display(df.head())

print("\nPlant distribution:")
print(df["PLANT_ID"].value_counts())
```

```python

df_model=df_plant_level.copy()

df_model=df_model.sort_values("DATE_TIME").set_index("DATE_TIME")

full_timeline=pd.date_range(
    start=df_model.index.min(),
    end=df_model.index.max(),
    freq="15min"
)

df_model=df_model.reindex(full_timeline)

df_model.index.name="DATE_TIME"

print("Original records:",len(df_plant_level))
print("Expected timeline records:",len(df_model))
print("Missing timestamps:",df_model["AC_POWER"].isnull().sum())

display(df_model.head(10))
```

```python

df_model["AC_POWER_LAG_15"]=df_model["AC_POWER"].shift(1)

df_model["AC_POWER_LAG_30"]=df_model["AC_POWER"].shift(2)

df_model["AC_POWER_LAG_60"]=df_model["AC_POWER"].shift(4)

print("Lag features created!")

display(df_model[[
    "AC_POWER",
    "AC_POWER_LAG_15",
    "AC_POWER_LAG_30",
    "AC_POWER_LAG_60"
]].head(10))
```

```python

df_model["TARGET_AC_POWER"]=df_model["AC_POWER"].shift(-1)

print("Future target created!")

display(df_model[[
    "AC_POWER",
    "TARGET_AC_POWER"
]].head(10))
```

```python

df_model["HOUR"]=df_model.index.hour

df_model["MINUTE"]=df_model.index.minute

df_model["TIME_OF_DAY"]=(
    df_model.index.hour*60+df_model.index.minute
)

df_model["HOUR_SIN"]=np.sin(
    2*np.pi*df_model["TIME_OF_DAY"]/1440
)

df_model["HOUR_COS"]=np.cos(
    2*np.pi*df_model["TIME_OF_DAY"]/1440
)

print("Time features created!")

display(df_model[[
    "HOUR",
    "MINUTE",
    "HOUR_SIN",
    "HOUR_COS"
]].head(10))
```

```python

feature_columns=[
    "AC_POWER",
    "DC_POWER",
    "AC_POWER_LAG_15",
    "AC_POWER_LAG_30",
    "AC_POWER_LAG_60",
    "IRRADIATION",
    "AMBIENT_TEMPERATURE",
    "MODULE_TEMPERATURE",
    "HOUR_SIN",
    "HOUR_COS"
]

target_column="TARGET_AC_POWER"

model_data=df_model[
    feature_columns+[target_column]
].copy()

print("Feature columns:",len(feature_columns))
print("Target column:",target_column)
```

```python

model_data=model_data.replace(
    [np.inf,-np.inf],
    np.nan
)

valid_data=model_data.dropna(
    subset=feature_columns+[target_column]
).copy()

print("Total timeline records:",len(model_data))
print("Valid forecasting samples:",len(valid_data))
print("Removed invalid samples:",len(model_data)-len(valid_data))

print("\nMissing values after cleaning:")
print(valid_data.isnull().sum())

print("\nFirst five valid samples:")
display(valid_data.head())
```

```python

sample_times=valid_data.index[:5]

for timestamp in sample_times:
    print(
        "Input time:",timestamp,
        "| Target time:",timestamp+pd.Timedelta(minutes=15)
    )
```

```python
valid_data=valid_data.sort_index()

n=len(valid_data)
train_end=int(n*0.80)
val_end=train_end+int(n*0.05)

train_data=valid_data.iloc[:train_end]
val_data=valid_data.iloc[train_end:val_end]
test_data=valid_data.iloc[val_end:]

X_train=train_data[feature_columns]
y_train=train_data[target_column]

X_val=val_data[feature_columns]
y_val=val_data[target_column]

X_test=test_data[feature_columns]
y_test=test_data[target_column]

print("Training samples:",len(train_data))
print("Validation samples:",len(val_data))
print("Testing samples:",len(test_data))

print("\nTraining period:",train_data.index.min(),"to",train_data.index.max())
print("Validation period:",val_data.index.min(),"to",val_data.index.max())
print("Testing period:",test_data.index.min(),"to",test_data.index.max())
```

```python
from pygam import LinearGAM,s
```

```python
gam=LinearGAM(
    s(0)+s(1)+s(2)+s(3)+s(4)+s(5)+s(6)+s(7)+s(8)+s(9)
)

gam.gridsearch(X_train.values,y_train.values)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import numpy as np

gam_val_pred=gam.predict(X_val.values)

gam_rmse=np.sqrt(mean_squared_error(y_val,gam_val_pred))
gam_mae=mean_absolute_error(y_val,gam_val_pred)
gam_r2=r2_score(y_val,gam_val_pred)

print("GAM Validation Results")
print("RMSE:",gam_rmse)
print("MAE:",gam_mae)
print("R2 Score:",gam_r2)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import SplineTransformer,StandardScaler
from sklearn.linear_model import Ridge
```

```python
spline_model=Pipeline([
    ("spline",SplineTransformer(n_knots=5,degree=3,knots="quantile")),
    ("scaler",StandardScaler()),
    ("ridge",Ridge(alpha=1.0))
])

spline_model.fit(X_train,y_train)
```

```python
spline_val_pred=spline_model.predict(X_val)

spline_rmse=np.sqrt(mean_squared_error(y_val,spline_val_pred))
spline_mae=mean_absolute_error(y_val,spline_val_pred)
spline_r2=r2_score(y_val,spline_val_pred)

print("Adaptive Spline Basis Validation Results")
print("RMSE:",spline_rmse)
print("MAE:",spline_mae)
print("R2 Score:",spline_r2)
```

```python
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor,VotingRegressor
```

```python
rf_model=RandomForestRegressor(n_estimators=150,max_depth=12,min_samples_split=5,random_state=42,n_jobs=-1)

gb_model=GradientBoostingRegressor(n_estimators=150,learning_rate=0.05,max_depth=3,random_state=42)

ensemble_model=VotingRegressor([
    ("random_forest",rf_model),
    ("gradient_boosting",gb_model)
])

ensemble_model.fit(X_train,y_train)
```

```python
ensemble_val_pred=ensemble_model.predict(X_val)

ensemble_rmse=np.sqrt(mean_squared_error(y_val,ensemble_val_pred))
ensemble_mae=mean_absolute_error(y_val,ensemble_val_pred)
ensemble_r2=r2_score(y_val,ensemble_val_pred)

print("Ensemble Validation Results")
print("RMSE:",ensemble_rmse)
print("MAE:",ensemble_mae)
print("R2 Score:",ensemble_r2)
```

```python
gam_test_pred=gam.predict(X_test.values)
spline_test_pred=spline_model.predict(X_test)
ensemble_test_pred=ensemble_model.predict(X_test)
```

```python
def evaluate_model(y_true,y_pred):
    rmse=np.sqrt(mean_squared_error(y_true,y_pred))
    mae=mean_absolute_error(y_true,y_pred)
    r2=r2_score(y_true,y_pred)
    wape=np.sum(np.abs(y_true-y_pred))/np.sum(np.abs(y_true))*100
    accuracy=100-wape
    return rmse,mae,r2,wape,accuracy

gam_test_metrics=evaluate_model(y_test.values,gam_test_pred)
spline_test_metrics=evaluate_model(y_test.values,spline_test_pred)
ensemble_test_metrics=evaluate_model(y_test.values,ensemble_test_pred)

test_results=pd.DataFrame({
    "Model":["GAM","Adaptive Spline","Ensemble"],
    "RMSE":[gam_test_metrics[0],spline_test_metrics[0],ensemble_test_metrics[0]],
    "MAE":[gam_test_metrics[1],spline_test_metrics[1],ensemble_test_metrics[1]],
    "R2":[gam_test_metrics[2],spline_test_metrics[2],ensemble_test_metrics[2]],
    "WAPE":[gam_test_metrics[3],spline_test_metrics[3],ensemble_test_metrics[3]],
    "Forecast_Accuracy":[gam_test_metrics[4],spline_test_metrics[4],ensemble_test_metrics[4]]
})

display(test_results.round(4))
```
